## Kruskal's Algorithm

Kruskal's algorithm is a greedy algorithm that finds a Minimum Spanning Tree (MST) for a weighted undirected graph.

### How it works:
1. **Sort** all the edges in non-decreasing order of their weight.
2. **Pick** the smallest edge. Check if it forms a cycle with the spanning tree formed so far. If a cycle is not formed, include this edge. Else, discard it.
3. **Repeat** step 2 until there are `(V-1)` edges in the spanning tree, where `V` is the number of vertices in the graph.

We use a **Disjoint Set Union (DSU)** (or Union-Find) data structure to efficiently detect cycles.
```

````python
class DisjointSet:
    def __init__(self, n):
        self.parent = list(range(n))
        self.rank = [0] * n

    def find(self, i):
        if self.parent[i] == i:
            return i
        # Path compression
        self.parent[i] = self.find(self.parent[i])
        return self.parent[i]

    def union(self, i, j):
        root_i = self.find(i)
        root_j = self.find(j)

        if root_i != root_j:
            # Union by rank
            if self.rank[root_i] < self.rank[root_j]:
                self.parent[root_i] = root_j
            elif self.rank[root_i] > self.rank[root_j]:
                self.parent[root_j] = root_i
            else:
                self.parent[root_j] = root_i
                self.rank[root_i] += 1
            return True
        return False


def kruskal(num_vertices, edges):
    # Sort edges based on their weight
    edges = sorted(edges, key=lambda item: item[2])

    ds = DisjointSet(num_vertices)
    mst = []
    mst_weight = 0

    for u, v, weight in edges:
        # If including this edge doesn't cause a cycle
        if ds.union(u, v):
            mst.append((u, v, weight))
            mst_weight += weight

        # Stop if we have V-1 edges
        if len(mst) == num_vertices - 1:
            break

    return mst, mst_weight

# Example usage:
# Graph representation: (u, v, weight)
edges = [
    (0, 1, 2),
    (0, 3, 6),
    (1, 2, 3),
    (1, 3, 8),
    (1, 4, 5),
    (2, 4, 7),
    (3, 4, 9)
]
num_vertices = 5

mst, total_weight = kruskal(num_vertices, edges)

print("Edges in the Minimum Spanning Tree:")
print("Edge \tWeight")
for u, v, weight in mst:
    print(f"{u} - {v} \t {weight}")
print(f"Total weight of MST: {total_weight}")
````

Read Me
- **Objective:** Finds the Minimum Spanning Tree (MST) of a weighted undirected graph by greedily choosing the smallest edges.
- **Data Structure:** Uses a Disjoint Set Union (DSU) with path compression and union-by-rank to quickly check for cycles in $O(E \log V)$ time.
- **Output:** Returns a list of edges forming the tree and the total minimum spanning weight (evaluated here as 16).
