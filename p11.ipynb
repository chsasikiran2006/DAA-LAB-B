Floyd-Warshall algorithm

```python
INF = float('inf')

def floyd_warshall(graph):
    """
    Finds all-pairs shortest paths using the Floyd-Warshall algorithm.

    Args:
        graph: A 2D list (adjacency matrix) where graph[i][j] is the weight
               of the edge from i to j, or INF if no direct edge exists.
    Returns:
        A 2D list containing the shortest distances between every pair of vertices.
    """
    num_vertices = len(graph)
    # Create a copy of the graph to store shortest distances
    dist = [row[:] for row in graph]

    # Try every vertex 'k' as an intermediate node
    for k in range(num_vertices):
        # Pick all vertices as source 'i' one by one
        for i in range(num_vertices):
            # Pick all vertices as destination 'j' for the above source
            for j in range(num_vertices):
                # If vertex k is on the shortest path from i to j, update dist[i][j]
                if dist[i][k] != INF and dist[k][j] != INF:
                    dist[i][j] = min(dist[i][j], dist[i][k] + dist[k][j])

    # Optional: Check for negative weight cycles
    for i in range(num_vertices):
        if dist[i][i] < 0:
            raise ValueError("Graph contains a negative weight cycle")

    return dist

# --- Example Usage ---
if __name__ == "__main__":
    # 4-vertex weighted directed graph
    # graph[i][j] represents edge weight from vertex i to vertex j
    graph = [
        [0,   3,   INF, 7],
        [8,   0,   2,   INF],
        [5,   INF, 0,   1],
        [2,   INF, INF, 0]
    ]

    shortest_paths = floyd_warshall(graph)

    print("Shortest distance matrix:")
    for row in shortest_paths:
        print([val if val != INF else "INF" for val in row])
```
